# **5. Revisión y tratamiento de tipo de dato**
El objetivo de esta sección es transformar las variables a formato númerico.

# Variable antigua_ne


Indica si la antigüedad de la vivienda no fue especificada.

Se identificó que si existen valores no nulos en antigua_ne, deberia de haber en antiguedad. Sin embargo, por algun motivo no se identifican como nulos.



```python
# Para el tratamiento de antiguedad Y antigua_ne
# Se crea una variable bandera y se convierte antiguedad a numérica.

# Crear bandera:
# 1 = No se especificó la antigüedad
# 0 = Sí se especificó la antigüedad

df_modelo['antiguedad_no_especificada'] = np.where(
    df_modelo['antigua_ne'].notna(),
    1,
    0
)

# Limpiar posibles espacios o códigos especiales en antiguedad
df_modelo['antiguedad'] = (
    df_modelo['antiguedad']
    .astype(str)
    .str.strip()
    .replace(['', ' ', '&', 'N', 'nan', 'NaN', '<NA>'], np.nan)
)

# Convertir antiguedad a número
df_modelo['antiguedad'] = pd.to_numeric(
    df_modelo['antiguedad'],
    errors='coerce'
)

# Revisar nulos reales después de la conversión
print('Nulos reales en antiguedad después de limpiar:')
print(df_modelo['antiguedad'].isna().sum())

print('\nCasos con antigüedad no especificada:')
print(df_modelo['antiguedad_no_especificada'].sum())
```

Después de limpiar y convertir `antiguedad` a formato numérico, se encontraron 12,278 valores faltantes, los cuales coinciden con los casos registrados en `antigua_ne`.

Para conservar estos registros se asignó el valor `-1` a los casos no especificados. Este valor se utiliza como código interno, ya que no pertenece al rango válido de antigüedad de la vivienda.

Adicionalmente, se creó la variable `antiguedad_no_especificada`, donde `1` indica que la antigüedad no fue reportada y `0` indica que sí fue especificada.

Despues se eliminó `antigua_ne`, debido a que su información quedó representada en la nueva variable bandera.


```python
# Reemplazar nulos de antiguedad con -1
df_modelo['antiguedad'] = df_modelo['antiguedad'].fillna(-1)

# Eliminar antigua_ne porque ya fue transformada en una variable útil
df_modelo = df_modelo.drop(columns=['antigua_ne'])
```

Finalmente, se realizaron las validaciones correspondientes para asegurarse que el proceso se haya realizado correctamente.

```python
# Validaciones
print('Validaciones')

# Verificar si la variable antigua_ne fue eliminada
print('\n¿La columna antigua_ne sigue en el dataset?', 'antigua_ne' in df_modelo.columns)

# Verificar nulos en antiguedad
print('\nNulos en antiguedad:', df_modelo['antiguedad'].isna().sum())

# Verificar cantidad de valores -1
print('\nCasos con antiguedad = -1:', (df_modelo['antiguedad'] == -1).sum())

# Verificar cantidad de casos marcados como no especificados
print('\nCasos con antiguedad_no_especificada = 1:', df_modelo['antiguedad_no_especificada'].sum())

# Verificar valores únicos de la bandera
print('\nValores únicos en antiguedad_no_especificada:', df_modelo['antiguedad_no_especificada'].unique())

# Verificar rango de antiguedad
print('\nValor mínimo y máximo de antiguedad:', df_modelo['antiguedad'].min(), ',', df_modelo['antiguedad'].max())

# Se filtran únicamente las variables que presentan valores nulos.
print('\n\nRegistros con valores nulos: ')
df_modelo.isnull().sum()[df_modelo.isnull().sum() > 0]
```
